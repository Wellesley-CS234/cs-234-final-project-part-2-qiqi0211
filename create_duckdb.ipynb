#### Notebook to create a duckdb database

```python
import duckdb
```

```python
CSV = "data/final.csv"

output_duckdb_path = "final.duckdb"
conn = duckdb.connect(database=output_duckdb_path)
```

```python
import time
start = time.time() # moment I started to run this cell

conn = duckdb.connect(database=output_duckdb_path)
conn.sql(f"""
    CREATE OR REPLACE TABLE wiki_pageviews AS
    SELECT
        date,
        qid,
        article,
        pageviews,
        language_full,
        truth,
        period
    FROM read_csv_auto(
        '{CSV}',
        SAMPLE_SIZE = -1,
        delim = ',',
        names = [
            'date',
            'qid',
            'article',
            'pageviews',
            'language_full',
            'predicted_label',
            'score',
            'truth',
            'period'
        ]
    )
""")
end = time.time() # moment the conversion was complete
totalTime = end - start
print(f"Conversion took {totalTime} seconds.")
```

```python
row_count = conn.sql("SELECT COUNT(*) FROM wiki_pageviews;").fetchone()[0]
```

```python
print(f"This table has {row_count} rows.")
```
